```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv(r"C:\Users\user\4ni25cs079\archive (2)\Employee_Salaries.csv")

print(df.columns)

# Seaborn Histogram
sns.histplot(data=df, x="Base_Salary", bins=10)

plt.title("Distribution of Base Salary")
plt.xlabel("Base Salary")
plt.ylabel("Frequency")
plt.show()


# -------------------------------
# Matplotlib Histogram
# -------------------------------

plt.hist(df["Base_Salary"], bins=10)

plt.title("Distribution of Base Salary")
plt.xlabel("Base Salary")
plt.ylabel("Frequency")
plt.show()

sns.histplot(data=df, x="Base_Salary", bins=10, kde=True)

plt.title("Distribution of Base Salary")
plt.xlabel("Base Salary")
plt.ylabel("Frequency")
plt.show()
```

```python

salary_data = df[
    ["Base_Salary", "Overtime_Pay", "Longevity_Pay"]
]

# Box plot
plt.boxplot(salary_data)

plt.title("Box Plot of Employee Salaries")
plt.xlabel("Salary Components")
plt.ylabel("Amount")
plt.xticks(
    [1, 2, 3],
    ["Base Salary", "Overtime Pay", "Longevity Pay"]
)

plt.show()

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
p = pd.read_csv(r"C:\Users\user\4ni25cs079\archive (3)\iris_synthetic_data.csv")
print(p.columns)
sns.scatterplot(
    data=p,
    x="petal length",
    y="petal width",
    hue="label"
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()


# -------------------------------
# Matplotlib Scatter Plot
# -------------------------------

plt.scatter(
    p["petal length"],
    p["petal width"]
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```
